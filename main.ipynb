```python
def Cleaned_Welder_Id(Val):
    # Checking if the the id is a 'NAN' value or None
    if pd.isna(Val) or Val is None:
        return np.nan
    
    # first we have to delect any extra spaces between
    Clean_Val = re.sub(r"[\s-]", "", str(Val)).upper()
    
    if not Clean_Val:
        return np.nan
    
    # we need to extract the 'W' character first
    digits = Clean_Val.lstrip("W")
    
    if len(digits) >= 3 and digits.startswith("0"):
        digits = digits[1:]
    
    if len(digits) == 1 :
        digits = digits.zfill(2)
        
    return f"W{digits}"

Welders_Ids = [
    "Welder_Id_One",
    "Welder_Id_Two",
]


for i in Welders_Ids:
    Welding_Log_Df[i] = Welding_Log_Df[i].apply(Cleaned_Welder_Id)
```

### Handling the missing values and duplicates
The dataset contains rows that appear duplicated.

After reviewing them with domain knowledge from the welding process, these records were intentionally retained because they represent welds that had not yet progressed through the execution stages. At the time of data extraction, operational fields such as Welding_Date, Welder_ID_One, Welder_ID_Two, ... were still empty, causing multiple records to appear identical

```python
Welding_Log_Df[Welding_Log_Df.duplicated(keep=False)]
```

For joints where the welding process has not been performed yet (Welding_Date is null), all test results (PT,Fit_Up, VT, and PAUT) default to 'Pending'.

```python
Test_Clms = [
    "Fit_Up",
    "Vt",
    "Pt",
    "Rt",
    "Paut_Report"
]


for clm in Test_Clms:
    Welding_Log_Df.loc[Welding_Log_Df["Welding_Date"].isnull(),clm] = "PENDING"
    
# We check if we have the report of each Test after the Welding if not we give a value of "NO REPORT"    
for clm in Test_Clms:
    Welding_Log_Df.loc[Welding_Log_Df[clm].isnull(), clm] = "NO REPORT"
```

### Standardizing Inspection Results
we generated standardized inspection result status columns (`Fit_Up_Result`, `Vt_Result`, `Pt_Result`, `Paut_Result`, `Rt_Result`) based on raw status entries while preserving the original columns:

* **`REJECTED`**: Assigned to joints marked as `"NO REPORT"`.
* **`Pending`**: Maintained for joints where welding/testing has not occurred yet.
* **`ACCEPTED`**: Default classification assigned to all successfully passed inspection records.

```python

mapping = {
    "Fit_Up": "Fit_Up_Result",
    "Vt": "Vt_Result",
    "Pt": "Pt_Result",
    "Paut_Report": "Paut_Result",
    "Rt" : "Rt_Result",
}

for Test_col, Result_col in mapping.items():
    conditions = [
        Welding_Log_Df[Test_col] == "NO REPORT",
        Welding_Log_Df[Test_col] == "PENDING",
    ]
    choices = ["REJECTED", "PENDING"]

    Welding_Log_Df[Result_col] = np.select(
        conditions, choices, default="ACCEPTED"
    )
```

### Primary Key Generation (`Weld_Id`)

Because `Joint_N` is non-unique across different Isometric drawings (`Iso_N`), we created a composite primary key **`Weld_Id`** by concatenating `Iso_N` and `Joint_N` (`Iso_N_Joint_N`).

```python
Welding_Log_Df["Weld_Id"] = (
    Welding_Log_Df["Iso_N"].astype(str)
    + "_"
    + Welding_Log_Df["Joint_N"].astype(str)
)
```
