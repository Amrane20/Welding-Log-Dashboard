```python
Welding_Log_Df.columns = (
    Welding_Log_Df.columns.str.strip().str.replace(" ", "_").str.replace("-", "_").str.title()
)
```

```python
Welding_Log_Df["Diameter"].unique()
```

```python
Welding_Log_Df["Diameter"] = Welding_Log_Df["Diameter"].replace('FW-11', 400).fillna(400)
```

```python
Welding_Log_Df["Diameter"].unique()
```

```python
Welding_Log_Df["Diameter"] = Welding_Log_Df["Diameter"].astype(int)
```

```python
df = Welding_Log_Df[Welding_Log_Df["Rev"].isnull()]
df.shape
```

```python
Welding_Log_Df["Rev"] =  Welding_Log_Df["Rev"].fillna(0)
Welding_Log_Df["Rev"] = Welding_Log_Df["Rev"].astype(int)
```

```python
Welding_Log_Df.columns
```

```python

Welding_Log_Df["Line_Class"] = Welding_Log_Df["Line_Class"].fillna("CD2")
```

```python
Welding_Log_Df[Welding_Log_Df["Joint_Type"] == "BW"].shape
```

```python
Welding_Log_Df["Joint_Type"].unique()
```

```python
Welding_Log_Df["Joint_Type"] = Welding_Log_Df["Joint_Type"].replace("FW-9", "FW").replace("9,53", "BW").replace(9.5,"BW").fillna("BW")
```
